# feature Enginnering

```python
# converting the date of journey from a object to date formatdf
# hear we were creating the date and month and Year features using the date_of_journey and for spliting wee nedd str.split('/') 
# we were spliting and seprating the date and month and year based on their index like[0]
# we can also convert it into predefined date or time feature , but it helps in forecasting tym
df['Date'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.info()
```

```python
# converting them into int type from object type
df['Date'] = df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year']= df['Year'].astype(int)
```

```python
df.info()
```

```python
# droping the date_of_journey, because we dont need it btw
df.drop('Date_of_Journey', axis = 1, inplace = True)
```

```python
# when we are doing feature engineering try to catch out simular kinds of fieds
# we were changing the arrival_time to int format , and we were also removing the date and month which are present in that arrival time, 
# we were using space to seprate the date here 
df['Arrival_Time'] = df['Arrival_Time'].str.split(' ').str[0]
# we can also use the lamba function also
```

```python
# here we were splitiing the hour and min
df['Arrival_hour'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_mins'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head()
```

```python
# here we were converting the Arrival_hour, Arrival_mins into int format
df['Arrival_hour'] = df['Arrival_hour'].astype(int)
df['Arrival_mins'] = df['Arrival_mins'].astype(int)
```

```python
# here we were dropping the Arrival_time feature from it
df.drop('Arrival_Time',axis = 1, inplace = True)
```

```python
## here we were also doing the same for dep_time also
df['Dep_hour'] = df['Dep_Time'].str.split(':').str[0]
df['Dep_mins'] = df['Dep_Time'].str.split(':').str[1]
```

```python
df.head()
```

```python
# here we were converting the Dep_hour, Dep_mins into int format
df['Dep_hour'] = df['Dep_hour'].astype(int)
df['Dep_mins'] = df['Dep_mins'].astype(int)
```

```python
# here we were aslo dropping the Dep_time here because we dont need it
df.drop('Dep_Time', axis = 1, inplace = True)
```

```python
# we were converting all the cateogrical values in to dummies
df=pd.get_dummies(df,columns=['Airline','Source','Destination','Additional_Info'], drop_first= True, dtype = int)
```

```python
df.head()
```

```python
# at final to make this data model ready lest make feature scalling 
#The tilde operator (~) acts as a logical NOT.
#This filters df to keep only the rows where the Purchase column is NOT missing. These rows are saved into a new DataFrame called df_train
df_train = df[~df['Price'].isnull()]
#This filters df to find all rows where the Purchase column contains missing (NaN) values. These rows are saved into a new DataFrame called df_test
df_test = df[df['Price'].isnull()]
```

```python
Y = df_train['Price']
X = df_train.drop('Price', axis=1)
```

```python
# test and train data splitting
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.30,
    random_state=100
)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, Y_train)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

y_pred = model.predict(X_test)

print("MAE:", mean_absolute_error(Y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(Y_test, y_pred)))
print("R2:", r2_score(Y_test, y_pred))
```
